```python
from gtracr.trajectory import Trajectory

# initialize a cosmic ray trajectory that arrives at the horizon
# from the North at 100km above sea level at the Kamioka site with
# a rigidity of 40 GV:

traj = Trajectory(
    location_name="Kamioka",
    zenith_angle=90.0,
    azimuth_angle=0.0,
    particle_altitude=100.0,
    rigidity=40.0,
    bfield_type="dipole",
)

# evaluate the trajectory and get the trajectory data
trajectory_data = traj.get_trajectory(get_data=True)

# plot the results
from gtracr.plotting import plot_3dtraj

plot_3dtraj([trajectory_data], mpl=False, show_plot=True)
```
